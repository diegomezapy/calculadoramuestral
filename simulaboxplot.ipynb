```
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import ipywidgets as widgets
from ipywidgets import interact

# Función para generar y modificar los datos con los nuevos parámetros
def generate_data(mean, std_dev, n, outlier_factor, num_outliers):
    # Generar datos normales
    data = np.random.normal(loc=mean, scale=std_dev, size=n)

    # Añadir valores atípicos
    outliers = np.random.uniform(low=mean + outlier_factor * std_dev,
                                 high=mean + outlier_factor * std_dev * 2,
                                 size=num_outliers)

    # Combinar los datos normales con los valores atípicos
    data_with_outliers = np.concatenate([data, outliers])
    return data_with_outliers

# Función para graficar el tallo y hojas y la caja de bigotes
def update(mean, std_dev, outlier_factor, num_outliers):
    # Generar los nuevos datos con los parámetros modificados
    data_with_outliers = generate_data(mean, std_dev, n, outlier_factor, num_outliers)

    # Crear DataFrame de los datos generados
    df = pd.DataFrame(data_with_outliers, columns=['Value'])

    # Limpiar los gráficos anteriores
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 12))

    # Calcular los cuartiles
    q1 = np.percentile(df['Value'], 25)
    q2 = np.percentile(df['Value'], 50)  # Mediana
    q3 = np.percentile(df['Value'], 75)

    # Crear gráfico tipo tallo y hojas (Stem-and-Leaf plot)
    sns.histplot(df['Value'], kde=False, ax=ax1, color='#4B8BBE', bins=30)
    ax1.set_title('Gráfico de Tallo y Hojas', fontsize=20, fontweight='bold', color='#2F4F4F')
    ax1.set_xlabel('Valor', fontsize=14, fontweight='bold', color='#2F4F4F')
    ax1.set_ylabel('Frecuencia', fontsize=14, fontweight='bold', color='#2F4F4F')

    # Añadir una línea vertical para el valor promedio
    mean_value = np.mean(df['Value'])
    ax1.axvline(mean_value, color='black', linestyle='--', linewidth=2)
    ax1.text(mean_value + 0.5, ax1.get_ylim()[1] * 0.9, f'Mean: {mean_value:.2f}',
             color='black', fontsize=12, fontweight='bold', va='center')

    # Ajustamos la escala de Y para mostrar mejor la distribución en logaritmos
    ax1.set_yscale('log')  # Ajustamos la escala si es necesario

    # Fijar límites de los ejes X de manera coherente entre los dos gráficos
    ax1.set_xlim(-100, 200)

    # Graficar la caja de bigotes
    sns.boxplot(x=df['Value'], ax=ax2, color='#FFD700', linewidth=2)
    ax2.set_title('Caja de Bigotes', fontsize=20, fontweight='bold', color='#2F4F4F')
    ax2.set_xlabel('Valor', fontsize=14, fontweight='bold', color='#2F4F4F')
    ax2.set_ylabel('Distribución', fontsize=14, fontweight='bold', color='#2F4F4F')

    # Añadir etiquetas para los cuartiles
    ax2.text(q1-30, 1, f'Q1: {q1:.2f}', color='black', fontsize=12, fontweight='bold')
    ax2.text(q2, 1, f'Mediana: {q2:.2f}', color='black', fontsize=12, fontweight='bold')
    ax2.text(q3+30, 1, f'Q3: {q3:.2f}', color='black', fontsize=12, fontweight='bold')

    # Mejorar bordes de los gráficos
    for box in ax2.artists:
        box.set_edgecolor('#FF4500')
        box.set_linewidth(3)

    # Fijar límites de los ejes X para el boxplot, asegurando que coincidan con el tallo y hojas
    ax2.set_xlim(-100, 200)

    # Ajustar la escala de Y del boxplot para evitar que se vea aplastado
    #ax2.set_ylim(df['Value'].min() - 10, df['Value'].max() + 10)  # Ajustamos los límites de Y

    # Ajustar el layout
    plt.tight_layout(pad=3)
    plt.show()

# Parámetros iniciales
n = 1000  # Número de datos
initial_mean = 50
initial_std_dev = 10
initial_outlier_factor = 3
initial_num_outliers = 20

# Crear los controles deslizantes (sliders)
mean_slider = widgets.FloatSlider(value=initial_mean, min=0, max=100, step=1, description="Media", style={'description_width': 'initial'})
std_dev_slider = widgets.FloatSlider(value=initial_std_dev, min=1, max=30, step=1, description="Desviación Estándar", style={'description_width': 'initial'})
outlier_slider = widgets.FloatSlider(value=initial_outlier_factor, min=1, max=10, step=0.1, description="Factor Atípico", style={'description_width': 'initial'})
outlier_count_slider = widgets.IntSlider(value=initial_num_outliers, min=0, max=100, step=1, description="Número de Atípicos", style={'description_width': 'initial'})

# Usar interact para crear la interactividad
interact(update,
         mean=mean_slider,
         std_dev=std_dev_slider,
         outlier_factor=outlier_slider,
         num_outliers=outlier_count_slider)
```
